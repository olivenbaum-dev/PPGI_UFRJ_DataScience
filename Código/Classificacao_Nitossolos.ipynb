# ADASYN

Grupo 1 - colunas: subordem1, gdegrupo1, area_km2

````python
from imblearn.over_sampling import RandomOverSampler
from sklearn.model_selection import train_test_split

df_clean = df.dropna(subset=['ordem1']).copy() #Limpeza de entradas nulas na coluna ordem1

y = df_clean['ordem1']

features_colunas_g1 = ['subordem1', 'gdegrupo1', 'area_km2'] # Ajuste com as colunas que seu modelo usará
X = df_clean[features_colunas_g1]

X = pd.get_dummies(X, drop_first=True) # Transformação de variáveis categóricas em numéricas (One-Hot Encoding)

# Divisão do modelo em Treinamento e Testes antes da aplicação da tecnica Adasyn - Fundamental para evitar overfitting
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y)

print(f"Antes do Over-sampling (Treino) - Qtd Nitossolos: {sum(y_train == 'NITOSSOLOS')}")

# Aplicação do RandomOverSampler no conjunto de TREINO da classe minoritária (Nitossolos)
ros = RandomOverSampler(random_state=42)
X_train_res, y_train_res = ros.fit_resample(X_train, y_train)

print(f"Depois do Over-sampling (Treino) - Qtd Nitossolos aproximada: {sum(y_train_res == 'NITOSSOLOS')}")
````

````python
# Geração da matriz de confusão do Adasyn G1

from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, confusion_matrix

modelo_solos = RandomForestClassifier(n_estimators=100, random_state=42)

modelo_solos.fit(X_train_res, y_train_res)

y_pred = modelo_solos.predict(X_test)

# Relatório de classificação completo
print(classification_report(y_test, y_pred))

# Matriz de Confusão para ver onde o modelo está errando
conf_matrix = confusion_matrix(y_test, y_pred)

plt.figure(figsize=(10, 8))
sns.heatmap(conf_matrix, annot=True, fmt='d', cmap='Blues',
            xticklabels=modelo_solos.classes_,
            yticklabels=modelo_solos.classes_)
plt.ylabel('Classe Real')
plt.xlabel('Classe Preditada pelo Modelo')
plt.title('Matriz de Confusão - Classificação de Solos - Adasyn - G1')
plt.show()
````

````python
# Metricas do Nitossolos

report_dict = classification_report(y_test, y_pred, output_dict=True)

classe_alvo = 'NITOSSOLOS'
if classe_alvo in report_dict:
    precisao_nito = report_dict[classe_alvo]['precision']
    recall_nito = report_dict[classe_alvo]['recall']
    f1_nito = report_dict[classe_alvo]['f1-score']
    acuracia_geral = report_dict['accuracy'] # A acurácia é global do modelo

    print(f"\nMétricas Isoladas para {classe_alvo}:")
    print(f"Precisão:  {precisao_nito:.4f}")
    print(f"Recall:     {recall_nito:.4f}")
    print(f"F1-Score:   {f1_nito:.4f}")
    print(f"Acurácia Geral do Modelo: {acuracia_geral:.4f}")
else:
    print(f"\nClasse '{classe_alvo}' não encontrada. Verifique se o nome está correto (maiúsculas/minúsculas).")
````

# **SMOTE**

Grupo 1 - colunas: subordem1, gdegrupo1, area_km2

````python
from imblearn.over_sampling import SMOTE             # SMOTE
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, confusion_matrix
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df_clean = df.dropna(subset=['ordem1']).copy()

# variável alvo
y = df_clean['ordem1']

features_colunas = ['subordem1', 'gdegrupo1', 'area_km2']

X = df_clean[features_colunas]

# Transformação de variáveis categóricas
X = pd.get_dummies(X, drop_first=True)

# Divisão do dataset em Treino e Teste
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y)

print(f"Antes do SMOTE (Treino) - Qtd Nitossolos: {sum(y_train == 'NITOSSOLOS')}")

# Aplicar o SMOTE clássico com k_neighbors menor devido ao tamanho da classe
smote = SMOTE(random_state=42, k_neighbors=2) # Reduzido k_neighbors para 2
X_train_res, y_train_res = smote.fit_resample(X_train, y_train)

print(f"Depois do SMOTE (Treino) - Qtd Nitossolos: {sum(y_train_res == 'NITOSSOLOS')}\n")

# Inicio e treinamento do modelo ajustado para evitar overfitting
modelo_solos = RandomForestClassifier(
    n_estimators=200,
    max_depth=15,          # Impede que o modelo decore demais as classes majoritárias
    min_samples_leaf=2,    # Exige mais robustez nas regras de decisão
    random_state=42
)
modelo_solos.fit(X_train_res, y_train_res)

y_pred = modelo_solos.predict(X_test)

print("--- NOVO RELATÓRIO DE CLASSIFICAÇÃO ---")
print(classification_report(y_test, y_pred))

# Matriz de Confusão
conf_matrix = confusion_matrix(y_test, y_pred)
plt.figure(figsize=(12, 9))
sns.heatmap(conf_matrix, annot=True, fmt='d', cmap='Blues',
            xticklabels=modelo_solos.classes_,
            yticklabels=modelo_solos.classes_)
plt.ylabel('Classe Real')
plt.xlabel('Classe Preditada pelo Modelo')
plt.title('Matriz de Confusão - Classificação de Solos - SMOTE - G1')
plt.show()
````

Grupo 2 - colunas: subordem1, gdegrupo1, area_km2, subordem2, simbolos, comp1

````python
# smote G2

from imblearn.over_sampling import SMOTE             # Mudamos para SMOTE E ADICIONAMOS NOVAS CLASSES
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, confusion_matrix
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns


df_clean = df.dropna(subset=['ordem1']).copy()

y = df_clean['ordem1']

features_colunas = [
    'subordem1', 'gdegrupo1',
    'ordem2', 'subordem2', 'gdegrupo2',   # <-- Novas colunas já incluídas
    'area_km2',
    'simbolos', 'comp1'
]
X = df_clean[features_colunas]


X = pd.get_dummies(X, drop_first=True)

# Divisão do dataset em Treino e Teste
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y)

print(f"Antes do SMOTE (Treino) - Qtd Nitossolos: {sum(y_train == 'NITOSSOLOS')}")


smote = SMOTE(random_state=42, k_neighbors=2) # Reduzido k_neighbors para 2
X_train_res, y_train_res = smote.fit_resample(X_train, y_train)

print(f"Depois do SMOTE (Treino) - Qtd Nitossolos: {sum(y_train_res == 'NITOSSOLOS')}\n")

# Inicio e treinamento do modelo ajustado para evitar overfitting
modelo_solos = RandomForestClassifier(
    n_estimators=200,
    max_depth=15,          # Impede que o modelo decore demais as classes majoritárias
    min_samples_leaf=2,    # Exige mais robustez nas regras de decisão
    random_state=42
)
modelo_solos.fit(X_train_res, y_train_res)

y_pred = modelo_solos.predict(X_test)

print("--- NOVO RELATÓRIO DE CLASSIFICAÇÃO ---")
print(classification_report(y_test, y_pred))

# Matriz de Confusão
conf_matrix = confusion_matrix(y_test, y_pred)
plt.figure(figsize=(12, 9))
sns.heatmap(conf_matrix, annot=True, fmt='d', cmap='Blues',
            xticklabels=modelo_solos.classes_,
            yticklabels=modelo_solos.classes_)
plt.ylabel('Classe Real')
plt.xlabel('Classe Preditada pelo Modelo')
plt.title('Nova Matriz de Confusão - Classificação de Solos - SMOTE- G2')
plt.show()
````
